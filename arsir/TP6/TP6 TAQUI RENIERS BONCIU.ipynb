```python
fr_frequencies = {'A' : 8.15,\
                  'B' : 0.97,\
                  'C' : 3.15,\
                  'D' : 3.73,\
                  'E' : 17.39,\
                  'F' : 1.12,\
                  'G' : 0.97,\
                  'H' : 0.85,\
                  'I' : 7.31,\
                  'J' : 0.45,\
                  'K' : 0.02,\
                  'L' : 5.69,\
                  'M' : 2.87,\
                  'N' : 7.12,\
                  'O' : 5.28,\
                  'P' : 2.80,\
                  'Q' : 1.21,\
                  'R' : 6.64,\
                  'S' : 8.14,\
                  'T' : 7.22,\
                  'U' : 6.38,\
                  'V' : 1.64,\
                  'W' : 0.03,\
                  'X' : 0.41,\
                  'Y' : 0.28,\
                  'Z' : 0.15}

letters = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
alphabet = letters + " "

def frequencies(message):
    f = {}

    # Cette fonction génère un dictionnaire qui donne la fréquence d'apparition de
    #lettre dans `message`, et retourne également la longueur effective du
    #message, c-à-d le nombre de caractères alphabétiques

    occurrences = {letter: 0 for letter in letters}
    len_message = 0

    for l in message:
        if l in letters:
            occurrences[l] += 1
            len_message += 1

    f = {}
    if len_message > 0:
        for letter in letters:
            f[letter] = (occurrences[letter] / len_message) * 100
    else:
        for letter in letters:
            f[letter] = 0.0

    return f, occurrences, len_message

def abs_distance(calculated_frequencies, target_frequencies, shift=0):
    # Cette fonction calcule la distance (somme des valeurs absolues des différences)
    #entre un tableau de fréquences donné d'une chaîne et celui du français
    #Optionnel : Un décalage peut être introduit pour calculer la distance entre un chiffrement de Cesar
    #par décalage d'un tableau de fréquences donné d'une chaine et celui du français

    distance = 0.0

    for i, letter in enumerate(letters):
        shifted_letter = letters[(i + shift) % 26]
        distance += abs(
            calculated_frequencies.get(shifted_letter, 0.0) -
            target_frequencies.get(letter, 0.0)
        )

    return distance

def coincidence_index(calculated_occurrences, len_message):
    # retourne l'indice de coïncidence
    # IC = sum(n_i * (n_i - 1)) / (N * (N - 1))
    if len_message < 2:
        return 0.0

    numerator = 0
    for n in calculated_occurrences.values():
        numerator += n * (n - 1)

    denominator = len_message * (len_message - 1)
    return numerator / denominator


msg_fq_test = input("Veuillez saisir un texte pour tester la class Decrypt")
fq, occ, len_msg = frequencies(msg_fq_test)
print("Les fréquences sont : \n {}".format(fq))

print("La distance par rapport au tableau de fréquences du français est : {}".format(abs_distance(fq, fr_frequencies)))

print(" l’indice de coïncidence est : {}".format(occ, len_msg))
```

```python
from random import shuffle
from collections import Counter
import unicodedata

class Subs :
  def __init__(self, key=None) :
    #Le contructeur initialise la clé (les substitutions) par le dictionnaire
    #donné en argument `key`, sinon il génère une clé aléatoirement
    #
    #Le dictionnaire donné peut ne pas inclure 26 entrées, celles non-montionnées restent inchangées.
    #exemple de key : {'A':S, 'T':O, 'Z':B ...}. S'il n'y pas la clé 'P' par exemple dans le dictionnaire,
    #la clé de l'objet cesar doit contenir l'entrée 'P':P.
    #
    #Vous pouvez construire la clé inverse pour faciliter le déchifferement.
    #
    self.key = {}
    self.key_inv = {}
    alphabet = [chr(i) for i in range(ord('A'), ord('Z')+1)]
    if key is None:
      shuffled = alphabet[:]
      shuffle(shuffled)
      for c, s in zip(alphabet, shuffled):
        self.key[c] = s
        self.key_inv[s] = c
    else:
      for c in alphabet:
        if key and c in key:
          self.key[c] = key[c]
          self.key_inv[key[c]] = c
        else:
          self.key[c] = c
          self.key_inv[c] = c


  def encode(self, plain):
    cipher = "".join([self.key[c] if c in self.key else c for c in plain])

    #Cette fonction utilise la clé pour
    #faire un chifferement par substitutions
    #
    #Le chiffrement de `plain` est retourné dans `cipher`

    return cipher

  def decode(self, cipher):
    plain = "".join([self.key_inv[c] if c in self.key_inv else c for c in cipher])

    #Cette fonction utilise la clé pour
    #faire un déchiffrement par substitutions
    #
    #Le déchiffrement de `cipher` est retourné dans `plain`

    return plain




def chosen_cipher(subs):
    assert isinstance(subs, Subs)
    #déduire la clé en utilisant subs.decode()
    #key =

    attack_word = [chr(i) for i in range(ord('A'), ord('Z')+1)] # alphabet
    attack_cipher = subs.decode(attack_word)

    key  = dict(sorted({plain: cipher for plain, cipher in zip(attack_cipher, attack_word)}.items()))

    print("la clé de chiffrement est : {}".format(key))

    return key

def chosen_plain(subs):
    assert isinstance(subs, Subs)
    #déduire la clé en utilisant subs.code() mais sans utiliser subs.decode()
    #keyabs =

    attack_word = [chr(i) for i in range(ord('A'), ord('Z')+1)] # alphabet
    attack_cipher = subs.encode(attack_word)

    key = dict(sorted({plain: cipher for plain, cipher in zip(attack_word, attack_cipher)}.items()))

    print("la clé de chiffrement est : {}".format(key))

    return key

def known_plain(plain, cipher):
    #cipher est le message obtenu en chiffrant plain
    #déduire une clé possible à partir de plain et cipher
    #keyabs =
    #

    key = dict(sorted({plain: cipher for plain, cipher in zip(plain, cipher)}.items()))
    # ajustement si le plain ne contient pas toutes les lettres de l'alphabet
    for c in [chr(i) for i in range(ord('A'), ord('Z')+1)]:
        if c not in key:
            key[c] = c
    # nettoyage des caractères non-alphabétiques
    key = {k: v for k, v in key.items() if k.isalpha() and v.isalpha()}

    print("la clé de chiffrement est : {}".format(key))
    return key


subs = Subs({'A': 'R', 'B': 'Y', 'C': 'B', 'D': 'Z', 'E': 'W', 'F': 'S', 'G': 'U', 'H': 'D', 'I': 'F', 'J': 'I', 'K': 'H', 'L': 'T', 'M': 'N', 'N': 'L', 'O': 'A', 'P': 'E', 'Q': 'K', 'R': 'G', 'S': 'C', 'T': 'P', 'U': 'Q', 'V': 'X', 'W': 'O', 'X': 'V', 'Y': 'M', 'Z': 'J'})
print(subs.key)
print(subs.key_inv)
print(subs.encode("LA CIGALE AYANT CHANTE TOUT LETE SE TROUVA FORT DEPOURVUE QUAND LA BISE FUT VENUE"))
print(subs.decode("TR BFURTW RMRLP BDRLPW PAQP TWPW CW PGAQXR SAGP ZWEAQGXQW KQRLZ TR YFCW SQP XWLQW"))
chosen_cipher(subs)
chosen_plain(subs)
known_plain("LA CIGALE AYANT CHANTE TOUT LETE SE TROUVA FORT DEPOURVUE QUAND LA BISE FUT VENUE", "TR BFURTW RMRLP BDRLPW PAQP TWPW CW PGAQXR SAGP ZWEAQGXQW KQRLZ TR YFCW SQP XWLQW")



```

# 4.   Vigenère

Adapter la classe `Cesar` pour chiffrer et déchiffrer un code de Vigenère.  Implémenter des méthodes d'attaque dans les cadres de texte chiffré choisi, texte clair choisi et texte clair connu.

```python
from random import randint

letters = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

def char_to_int(c):
    return letters.index(c)

def int_to_char(i):
    return letters[i % 26]

def key_to_shifts(key):
    key = key.replace(" ", "")
    return [char_to_int(c) for c in key]

def shifts_to_key(shifts):
    return "".join(int_to_char(s) for s in shifts)

def minimal_repeating_pattern(seq):
    n = len(seq)
    for p in range(1, n + 1):
        ok = True
        for i in range(n):
            if seq[i] != seq[i % p]:
                ok = False
                break
        if ok:
            return seq[:p]
    return seq

class Vigenere :
  def __init__(self, key=None) :
    if key is None:
        key_length = randint(3, 8)
        self.key = "".join(letters[randint(0, 25)] for _ in range(key_length))
    else:
        key = key.replace(" ", "")
        if len(key) == 0:
            raise ValueError("La clé ne doit pas être vide.")
        self.key = key

  def encode(self, plain):
    cipher = ""

    #Cette fonction utilise la clé pour
    #faire un chifferement de Vigenère
    #
    #Le chifferement de `plain` est retourné dans `cipher`

    key_shifts = key_to_shifts(self.key)
    j = 0

    for c in plain:
        if c == " ":
            cipher += " "
        else:
            shift = key_shifts[j % len(key_shifts)]
            cipher += int_to_char(char_to_int(c) + shift)
            j += 1

    return cipher

  def decode(self, cipher):
    plain = ""

    #Cette fonction utilise la clé pour
    #faire un déchiffrement de Vigenère
    #
    #Le déchiffrement de `cipher` est retourné dans `plain`

    key_shifts = key_to_shifts(self.key)
    j = 0

    for c in cipher:
        if c == " ":
            plain += " "
        else:
            shift = key_shifts[j % len(key_shifts)]
            plain += int_to_char(char_to_int(c) - shift)
            j += 1

    return plain

class DecryptVigenere :
  def chosen_cipher(self, vigenere):
    assert isinstance(vigenere, Vigenere)
    #déduire la clé en utilisant vigenere.decode()
    # On choisit un texte chiffré composé uniquement de A
    # decode(A) = A - clé, donc on peut retrouver la clé
    probe = "A" * 200
    decoded = vigenere.decode(probe)

    key_stream = [(-char_to_int(c)) % 26 for c in decoded]
    key_stream = minimal_repeating_pattern(key_stream)
    key = shifts_to_key(key_stream)

    print("la clé de chiffrement est : {}".format(key))
    return key

  def chosen_plain(self, vigenere):
    assert isinstance(vigenere, Vigenere)
    #déduire la clé en utilisant vigenere.code() mais sans utiliser vigenere.decode()
    # Si on chiffre AAAAA..., le résultat est directement la clé répétée
    probe = "A" * 200
    encoded = vigenere.encode(probe)

    key_stream = [char_to_int(c) for c in encoded]
    key_stream = minimal_repeating_pattern(key_stream)
    key = shifts_to_key(key_stream)

    print("la clé de chiffrement est : {}".format(key))
    return key

  def known_plain(self, plain, cipher):
    #cipher est le message obtenu en chiffrant plain
    #déduire la clé à partir de plain et cipher

    if len(plain) != len(cipher):
        raise ValueError("plain et cipher doivent avoir la même longueur.")

    key_stream = []

    for p, c in zip(plain, cipher):
        if p == " " and c == " ":
            continue
        if p == " " or c == " ":
            raise ValueError("Les espaces doivent être aux mêmes positions.")
        shift = (char_to_int(c) - char_to_int(p)) % 26
        key_stream.append(shift)

    key_stream = minimal_repeating_pattern(key_stream)
    key = shifts_to_key(key_stream)

    print("la clé de chiffrement est : {}".format(key))
    return key

# pour tester
vig = Vigenere("CLE")
cipher = vig.encode("BONJOUR A TOUS")
print(cipher)

plain = vig.decode(cipher)
print(plain)

vig = Vigenere("CLE")
attacker = DecryptVigenere()

attacker.chosen_plain(vig)
attacker.chosen_cipher(vig)

p = "BONJOUR A TOUS"
c = vig.encode(p)
attacker.known_plain(p, c)
```

## Test de Friedman
Si la longueur de la clé est `k`, alors pour tout `i`, le sous-message formé par les lettres aux positions
`i mod k` est chiffré avec un chiffrement de César. En particulier, son indice de coïncidence doit
être proche de 0.07 (et non de 0.03).

En utilisant ce critère, écrire une fonction qui détermine la longueur (la plus probable) de la
clé. Écrire ensuite une fonction qui calcule la clé (la plus probable).

```python
from collections import Counter

def _clean_text(message):
    return "".join(c for c in message.upper() if c.isalpha())

def _split_by_key_length(cipher, length_key):
    cipher = _clean_text(cipher)
    return [cipher[i::length_key] for i in range(length_key)]

def _coincidence_from_text(message):
    f, n = frequencies(message)
    if n <= 1:
        return 0.0

    occurrences = {}
    for letter in f:
        occurrences[letter] = f[letter] * n / 100.0

    return coincidence_index(occurrences, n)

def _best_caesar_shift(message):
    f, _ = frequencies(message)

    best_shift = 0
    best_dist = float("inf")

    for shift in range(26):
        d = abs_distance(f, fr_frequencies, shift)
        if d < best_dist:
            best_dist = d
            best_shift = shift

    return best_shift

def friedman_length(cipher):
    cipher = _clean_text(cipher)
    if len(cipher) < 2:
        return 1

    best_length = 1
    best_score = -1

    max_len = min(20, len(cipher))

    for length in range(1, max_len + 1):
        blocks = _split_by_key_length(cipher, length)
        scores = []

        for block in blocks:
            if len(block) > 1:
                scores.append(_coincidence_from_text(block))

        if len(scores) == 0:
            continue

        avg_score = sum(scores) / len(scores)

        if avg_score > best_score:
            best_score = avg_score
            best_length = length

    return best_length

def cipher_only(cipher, length_key):
    cipher_clean = _clean_text(cipher)
    blocks = _split_by_key_length(cipher_clean, length_key)

    shifts = []
    key = ""

    for block in blocks:
        shift = _best_caesar_shift(block)
        shifts.append(shift)
        key += chr(ord('A') + shift)

    plain = ""
    j = 0

    for c in cipher:
        if c.isalpha():
            x = ord(c.upper()) - ord('A')
            shift = shifts[j % length_key]
            plain += chr((x - shift) % 26 + ord('A'))
            j += 1
        else:
            plain += c

    return key, plain
```

Implémenter une fonction qui détermine la liste des (positions des) sous-mots d’au moins
`p` lettres apparaîssant plusieurs fois dans un message.

```python
from collections import Counter
from itertools import combinations

def repetitions(cipher, p):
    cipher = cipher.replace(" ", "")
    n = len(cipher)

    found = []

    for size in range(p, n + 1):
        seen = {}

        for i in range(n - size + 1):
            sub = cipher[i:i + size]
            if sub not in seen:
                seen[sub] = [i]
            else:
                seen[sub].append(i)

        for sub, positions in seen.items():
            if len(positions) >= 2:
                found.append((sub, positions))

    return found
```

La plupart de ces répétitions correspondent (sauf faux-positif) à des sous-mots du message en
clair qui ont été chiffrés avec les mêmes portions de la clé. Dans ce cas, l’écart entre leurs positions
est un multiple de la longueur de la clé. En déduire une fonction qui fournit la longueur (probable) de la clé.

```python
def bk_length(cipher):
    reps = repetitions(cipher, 3)

    if not reps:
        return 0

    divisor_count = Counter()

    for sub, positions in reps:
        for a, b in combinations(positions, 2):
            distance = b - a

            for d in range(2, distance + 1):
                if distance % d == 0:
                    divisor_count[d] += 1

    if not divisor_count:
        return 0

    probable_length = divisor_count.most_common(1)[0][0]
    return probable_length
```

# 5. Attaque par dictionnaire

Les attaques par fréquences ne fournissent pas directement la clé en général. En revanche,
elles fournissent une bonne estimation de à quoi ressemble la clé. On peut alors chercher des clés
« proches » et tester si elles décryptent le message.

Vous trouverez à l’adresse https://www.freelang.com/download/misc/liste_francais.zip
un fichier contenant tous les mots français. Commencez par faire un prétraitement qui enlève tous

Écrivez ensuite une fonction qui cherche **efficacement** si un mot est dans le dictionnaire, et une qui, étant donné un message, calcule la proportion de ses mots qui sont bien français.

```python
def cleanup(word):
    return ''.join(c for c in unicodedata.normalize('NFD', word) if unicodedata.category(c) != 'Mn' and c.isalpha()).upper()

dico = set(cleanup(word) for word in open("mots.txt", "r").read().splitlines())

def recherche(word):
    return word in dico

def proportion(cipher):
    word_split = cipher.split()
    total = len(word_split)
    found = sum(1 for word in word_split if recherche(cleanup(word)))
    return found / total if total > 0 else 0
```
