```python
from ortools.sat.python import cp_model
from ortools.linear_solver import pywraplp
```

```python
### Voitures (Nombres entiers) ###

model = cp_model.CpModel()

# variables
x = []
x.append(model.NewIntVar(0,1000000,f"x_{0}"))
x.append(model.NewIntVar(0,1000000,f"x_{1}"))

# Contraintes
model.Add(6*x[0]+5*x[1]<=6000)
model.Add(x[0]+2*x[1]<=1500)
model.Add(x[1]<=800)

# Fonction objectif
model.Maximize(5000*x[0]+4500*x[1])

# Solution
solver = cp_model.CpSolver()
status = solver.Solve(model)
if status == cp_model.OPTIMAL or status == cp_model.FEASIBLE:
    print("x_1 = " + str(solver.Value(x[0])))
    print("x_2 = " + str(solver.Value(x[1])))
    print("z = " + str(solver.ObjectiveValue()))
else:
     print('No solution')
```

```python
### Voitures (Nombres réels) ###

solver = pywraplp.Solver.CreateSolver("GLOP")

# variables
x = []
x.append(solver.NumVar(0,solver.infinity(),"x_0"))
x.append(solver.NumVar(0,solver.infinity(),"x_1"))

# Contraintes
solver.Add(6*x[0]+5*x[1]<=6000)
solver.Add(x[0]+2*x[1]<=1500)
solver.Add(x[1]<=800)

# Fonction objectif
solver.Maximize(5000*x[0]+4500*x[1])

# Solution
status = solver.Solve()
if status == pywraplp.Solver.OPTIMAL or status == pywraplp.Solver.FEASIBLE:
    print("x_1 = " + str(x[0].solution_value()))
    print("x_2 = " + str(x[1].solution_value()))
    print("z = " + str(solver.Objective().Value()))
else:
     print('No solution')
```

```python
### Investissement financier (Nombres entiers) ###

model = cp_model.CpModel()

# variables
x = []
for i in range(5):
    x.append(model.NewBoolVar(f"x_{i}"))

# Contraintes
model.Add(75*x[0]+43*x[1]+56*x[2]+12*x[3]+31*x[4]<=100)

# Fonction objectif
c = [210,90,300,30,70]
model.Maximize(sum((c[i]*x[i]) for i in range(5)))

# Solution
solver = cp_model.CpSolver()
status = solver.Solve(model)
if status == cp_model.OPTIMAL or status == cp_model.FEASIBLE:
    for i in range(5):
        print(f"x_{i+1} = " + str(solver.Value(x[i])))
    print("z = " + str(solver.ObjectiveValue()))
else:
     print('No solution')
```

```python
### Investissement financier (Nombres réels) ###

solver = pywraplp.Solver.CreateSolver("GLOP")

# variables
x = []
for i in range(5):
    x.append(solver.NumVar(0,1,f"x_{i}"))

# Contraintes
solver.Add(75*x[0]+43*x[1]+56*x[2]+12*x[3]+31*x[4]<=100)

# Fonction objectif
c = [210,90,300,30,70]
solver.Maximize(sum((c[i]*x[i]) for i in range(5)))

# Solution
status = solver.Solve()
if status == pywraplp.Solver.OPTIMAL or status == pywraplp.Solver.FEASIBLE:
    for i in range(5):
        print(f"x_{i+1} = " + str(x[i].solution_value()))
    print("z = " + str(solver.Objective().Value()))
else:
     print('No solution')
```

```python
### Découpe de barres métalliques ###

model = cp_model.CpModel()

# variables
x = []
for i in range(15):
    x.append(model.NewIntVar(0,1000000,f"x_{i}"))

# Contraintes
model.Add(2*x[2]+2*x[4]+3*x[5]+x[8]+3*x[9]+5*x[11]+x[12]+2*x[13]+3*x[14]>=432)
model.Add(x[1]+2*x[3]+x[4]+x[6]+2*x[7]+3*x[10]+x[12]+2*x[13]+x[14]>=500)
model.Add(2*x[0]+x[1]+x[2]+2*x[6]+x[7]+2*x[8]+x[9]+x[12]>=400)

# Fonction objectif
c = [150, 150, 150, 150, 150, 150, 200, 200, 200, 200, 200, 200, 200, 200, 200]
model.Minimize(sum((c[i]*x[i]) for i in range(15)))

# Solution
solver = cp_model.CpSolver()
status = solver.Solve(model)
if status == cp_model.OPTIMAL or status == cp_model.FEASIBLE:
    for i in range(15):
        print(f"x_{i+1} = " + str(solver.Value(x[i])))
    print("z = " + str(solver.ObjectiveValue()))
else:
     print('No solution')
```
